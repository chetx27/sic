### 1. Load the Iris Dataset and Convert to NumPy Array

```python
import pandas as pd
import numpy as np

df = pd.read_csv('/content/Iris.csv')

# Extract the numeric columns (SepalLengthCm, SepalWidthCm, PetalLengthCm, PetalWidthCm)
iris_features = df.iloc[:, 1:5].values
print("Iris features shape:", iris_features.shape)
display(df.head())
```

### 2. Basic Properties & Indexing/Slicing
Let's check the ndim, size, and extract specific parts of the dataset.

```python
print("Dimensions:", iris_features.ndim)
print("Size (total elements):", iris_features.size)

# Slice the first 5 rows of the dataset
print("\nFirst 5 rows:\n", iris_features[:5])

# Find the argmax and argmin of the first feature column (SepalLengthCm)
sepal_length = iris_features[:, 0]
print("\nMax Sepal Length Index:", sepal_length.argmax())
print("Min Sepal Length Index:", sepal_length.argmin())
```

### 3. Matrix & Linear Algebra Operations
We can slice a square matrix (e.g., $2 \times 2$) from our dataset features to perform matrix multiplication, transpose, find the determinant, inverse, and eigenvalues.

```python
# Take a 2x2 matrix from the first row's Sepal and Petal measurements
matrix_iris = iris_features[0].reshape(2, 2)
print("Original 2x2 Matrix from first sample:\n", matrix_iris)

# 1. Transpose
print("\nTranspose:\n", matrix_iris.T)

# 2. Determinant
det = np.linalg.det(matrix_iris)
print("\nDeterminant:", det)

# 3. Inverse
try:
    inv = np.linalg.inv(matrix_iris)
    print("\nInverse:\n", inv)
except np.linalg.LinAlgError:
    print("\nMatrix is singular and cannot be inverted.")

# 4. Eigenvalues and Eigenvectors
eig_vals, eig_vecs = np.linalg.eig(matrix_iris)
print("\nEigenvalues:", eig_vals)
print("Eigenvectors:\n", eig_vecs)
```
